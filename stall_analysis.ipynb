# Gikomba Stall Performance Analysis
**Client:** Wanjiku (Gikomba Market)  
**Tools:** Core Python 3 (No external libraries)

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Metrics
### Approach
We use Python's built-in functions `sum()`, `len()`, `max()`, and `min()` to compute aggregate metrics across the 14-day sales list. Using `.index()`, we find the positions of the highest and lowest sales days to map them to their corresponding day name and determine whether they occurred in Week 1 (indices 0 to 6) or Week 2 (indices 7 to 13).

```python
# Task 1: Calculate core metrics
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Identify day positions
best_day_idx = daily_sales.index(highest_sales)
worst_day_idx = daily_sales.index(lowest_sales)

best_day_name = days[best_day_idx]
worst_day_name = days[worst_day_idx]

# Check week number (positions 0 to 6 are Week 1; 7 to 13 are Week 2)
best_day_week = 1 if best_day_idx < 7 else 2
worst_day_week = 1 if worst_day_idx < 7 else 2

# Print the results
print("Total Sales: KES", total_sales)
print("Average Daily Sales: KES", average_sales)
print("Highest Day:", best_day_name, f"(Week {best_day_week}) - KES", highest_sales)
print("Lowest Day:", worst_day_name, f"(Week {worst_day_week}) - KES", lowest_sales)
```

## Task 2: Weekly Totals and Week-on-Week Growth
### Approach
We define a function `week_totals(sales)` that uses Python list slicing (`[:7]` for Week 1 and `[7:]` for Week 2) to calculate and return `(week_1_total, week_2_total)` as a tuple. We include a docstring per assignment requirements. After unpacking the tuple, we calculate the week-on-week growth percentage using `((week_2 - week_1) / week_1) * 100`.

```python
# Task 2: Function for weekly totals and growth rate

def week_totals(sales):
    """
    Splits a 14-day sales list into two 7-day segments and returns their totals.

    Args:
        sales (list): A list of 14 numerical sales figures.

    Returns:
        tuple: A pair containing (week_1_total, week_2_total).
    """
    week_1_total = sum(sales[:7])
    week_2_total = sum(sales[7:])
    return (week_1_total, week_2_total)

# Call the function and unpack the tuple
week_1, week_2 = week_totals(daily_sales)

# Calculate week-on-week growth percentage
wow_growth = ((week_2 - week_1) / week_1) * 100

# Print the results
print("Week 1 Total: KES", week_1)
print("Week 2 Total: KES", week_2)
print(f"Week-on-Week Growth: {wow_growth:+.2f}%")
```

## Task 3: Day Classification
### Approach
We define `classify_day(amount, average)` with a docstring to categorize daily sales as 'Strong' (> 20% above average), 'Weak' (< 20% below average), or 'Normal'. We apply this function across all 14 days and tabulate the occurrences using a dictionary counting pattern (`counts[category] = counts.get(category, 0) + 1`).

```python
# Task 3: Day classification function and counting

def classify_day(amount, average):
    """
    Classifies a daily sales figure relative to the average.

    Args:
        amount (float or int): The daily revenue amount.
        average (float or int): The 14-day average daily revenue.

    Returns:
        str: 'Strong' if > 20% above average, 'Weak' if > 20% below average,
             otherwise 'Normal'.
    """
    if amount > 1.20 * average:
        return "Strong"
    elif amount < 0.80 * average:
        return "Weak"
    else:
        return "Normal"

# Apply classification to each day
day_classifications = [classify_day(val, average_sales) for val in daily_sales]

# Count the frequency of each category using the dictionary counting pattern
day_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
for status in day_classifications:
    day_counts[status] = day_counts.get(status, 0) + 1

# Print the counts
for category, count in day_counts.items():
    print(f"{category} Days: {count}")
```

## Task 4: Weekday vs. Weekend Analysis
### Approach
We separate weekday sales (Monday through Friday) from weekend sales (Saturday and Sunday) using index positions modulo 7 (`0–4` map to Mon–Fri, while `5–6` map to Sat–Sun). We compute the mean for both subsets and calculate the percentage premium of weekend sales relative to weekday sales using `((avg_weekend - avg_weekday) / avg_weekday) * 100`.

```python
# Task 4: Weekday vs Weekend Analysis

# Filter sales based on position in the week (0-4: Mon-Fri, 5-6: Sat-Sun)
weekday_sales = [daily_sales[i] for i in range(len(daily_sales)) if i % 7 < 5]
weekend_sales = [daily_sales[i] for i in range(len(daily_sales)) if i % 7 >= 5]

# Calculate averages
avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100

# Print results
print(f"Weekday Average (10 days): KES {avg_weekday:,.2f}")
print(f"Weekend Average (4 days):  KES {avg_weekend:,.2f}")
print(f"Weekend Difference: {weekend_diff_pct:+.2f}%")
```

## Task 5: Inventory Reorder Status and Margin Analysis
### Approach
We define a custom function `get_reorder_list(stock_data)` with a docstring to identify items where current inventory is at or below the minimum reorder threshold (`qty <= reorder`). Using a `for` loop with `.items()`, we iterate through each product to calculate its gross profit margin percentage using `((price - cost) / price) * 100`.

```python
# Task 5: Inventory reorder function and margin calculation

def get_reorder_list(stock_data):
    """
    Identifies inventory items whose quantity is at or below the reorder level.

    Args:
        stock_data (dict): Nested dictionary containing stock details per item.

    Returns:
        list: Names of products requiring immediate restocking.
    """
    reorder_needed = []
    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            reorder_needed.append(product)
    return reorder_needed

# Get list of products needing reorder
items_to_reorder = get_reorder_list(stock)

# Calculate gross margin percentage for each product
product_margins = {}
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    product_margins[product] = margin

# Print results
print("Products to restock immediately:", items_to_reorder)
print("\nGross Profit Margins per Product:")
for product, margin in product_margins.items():
    print(f"  {product}: {margin:.2f}%")
```

## Task 6: Formatted Stall Performance Report
### Approach
We compile all summary metrics, weekly figures, day classifications, and inventory statuses into a clean, aligned ASCII report. We use Python f-strings with field width specifiers (`<` for left-aligned, `>` for right-aligned) and currency formatting (`:,.2f`) to include thousands separators.

```python
# Task 6: Print Formatted Stall Performance Report

print("=" * 64)
print(" " * 19 + "STALL PERFORMANCE REPORT")
print(" " * 20 + "Gikomba Market Stall")
print("=" * 64)
print(f"{'Metric':<36} | {'Value':>23}")
print("-" * 64)
print(f"{'Total Revenue (14 Days)':<36} | KES {total_sales:>19,.2f}")
print(f"{'Daily Average Revenue':<36} | KES {average_sales:>19,.2f}")
print(f"{'Best Day Revenue':<36} | KES {highest_sales:>19,.2f}")
print(f"{'  -> Best Day Details':<36} | {best_day_name + ' (Week ' + str(best_day_week) + ')':>23}")
print(f"{'Lowest Day Revenue':<36} | KES {lowest_sales:>19,.2f}")
print(f"{'  -> Lowest Day Details':<36} | {worst_day_name + ' (Week ' + str(worst_day_week) + ')':>23}")
print("-" * 64)
print(f"{'Week 1 Total Revenue':<36} | KES {week_1:>19,.2f}")
print(f"{'Week 2 Total Revenue':<36} | KES {week_2:>19,.2f}")
print(f"{'Week-on-Week Growth Rate':<36} | {wow_growth:>22.2f}%")
print("-" * 64)
print(f"{'Weekday Average (Mon-Fri)':<36} | KES {avg_weekday:>19,.2f}")
print(f"{'Weekend Average (Sat-Sun)':<36} | KES {avg_weekend:>19,.2f}")
print(f"{'Weekend Difference vs Weekdays':<36} | {weekend_diff_pct:>22.2f}%")
print("-" * 64)
print(f"{'Performance Distribution':<36} | Strong: {day_counts['Strong']} | Norm: {day_counts['Normal']} | Weak: {day_counts['Weak']}")
print("=" * 64)
print(" " * 20 + "INVENTORY & MARGIN HEALTH")
print("-" * 64)
print(f"{'Product':<12} | {'Qty (kg)':>8} | {'Reorder':>8} | {'Status':<10} | {'Margin':>8}")
print("-" * 64)
for product, details in stock.items():
    status = "REORDER" if details["qty"] <= details["reorder"] else "OK"
    margin = product_margins[product]
    print(f"{product:<12} | {details['qty']:>8} | {details['reorder']:>8} | {status:<10} | {margin:>7.2f}%")
print("=" * 64)
```
